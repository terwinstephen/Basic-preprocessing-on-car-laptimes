```python
import pandas as pd
import matplotlib.pyplot as pl

#Displaying the first 5 values of the dataset
df=pd.read_excel('/home/dell/Documents/car.xlsx')
display(df.head())

#Analysing how many models each manufacturer has
m_counts=df['Manufacture'].value_counts()
pl.figure(figsize=(10,20))
m_counts.plot(kind='pie',color='violet')
pl.show()
pl.figure(figsize=(10,7))
m_counts.plot(kind='barh',color='violet')
pl.show()


#identifying outliers in price
pl.figure(figsize=(7,7))
pl.boxplot(df['price'])
pl.title("Price anomaly dectection")
pl.ylabel("price")
pl.show()

#deleting outliers in price
df2=df[df['price']>1000000].index
df.drop(df2,inplace=True)
display(df.head())

#identifying missing values in the dataset
print("missing values")
display(df.isnull().sum())
display(df.head())

#Deleting all models without laptimes
print("\033[31mDelete values without lap times\033[0m")
df_1=df.dropna(subset=['Lap Time'])
df=df_1
display(df.isnull().sum())
display(df.head())

#to quantify performance stats create a new column for power to weight ratio
df_2=df.assign(ratio=df['Hp']/df['Weight'])
df=df_2
display(df.head(30))

#plotting a bar graph to visualize all models and their p/w ratio
df['Model'] = df['Model'].astype(str)
pl.figure(figsize=(10,7))
pl.barh(df['Model'],df['ratio'])
pl.show()

#deleting colums without power or weight stats
df_3 = df.dropna(subset=['ratio'])
df=df_3
df = df.reset_index(drop=True)
display(df.head(30))
display(df.isnull().sum())


#plotting various scatterplots to visualize p/w, top speed and price stats
pl.figure(figsize=(7,7))
pl.scatter(df['price'],df['ratio'],s=200,color='#ffc500')
pl.title("Price to P/W ratio")
pl.xlabel("price")
pl.ylabel("Power to weight ratio")
pl.grid()
pl.show()

pl.figure(figsize=(7,7))
pl.scatter(df['price'],df['Top Speed'],s=200,color='#ff0000')
pl.title("Price to Top Speed")
pl.xlabel("price")
pl.ylabel("Top Speed")
pl.grid()
pl.show()


#filtering out various stats for customer ,raio(0.3<x<0.4),top speed(300<x)
df_4=df[df['ratio']<=0.30].index
df.drop(df_4,inplace=True)
df_5=df[df['ratio']>=0.40].index
df.drop(df_5,inplace=True)

#plotting new modifed graphs
pl.scatter(df['price'],df['ratio'],s=200,color='#ff6800')
pl.title("Price to P/W ratio")
pl.xlabel("price")
pl.ylabel("Power to weight ratio")
pl.grid()
pl.show()

#for top speed
df_6=df[df['Top Speed']<300].index
df.drop(df_6,inplace=True)

pl.figure(figsize=(7,7))
pl.scatter(df['price'],df['Top Speed'],s=200,color='#ff00a6')
pl.title("Price to Top Speed")
pl.xlabel("price")
pl.ylabel("Top Speed")
pl.grid()
pl.show()

#resetting index
df = df.reset_index(drop=True)

#assigning a new parameter Value for money (hp obtained for 1$)
df_7=df.assign(value=df['Hp']/df['price'])
df=df_7
display(df.head(10))

#plotting final graph to find out the best value for money car
pl.figure(figsize=(20,10))
pl.bar(df['Model'],df['value'],color='#8bf573')
pl.title("Most value")
pl.xlabel("Model")
pl.ylabel("value")
pl.grid()
pl.show()


```
